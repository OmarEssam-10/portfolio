```python
df['chol_group'] = pd.cut( df['chol'] , bins = [100,200,300,400,500, float('inf')] , 
                          labels = [ '100-199' , '200-299' , '300-399' , '400-499' , '500+' ] , right = False )

chol_group_analysis = df.groupby('chol_group' , observed = False).agg(total_patients = ('target' , 'count') , disease_rate = ('target' , 'mean'))

chol_group_analysis['disease_rate'] *=100

print(chol_group_analysis . round(2))
```

```python
df['trestbps_group'] = pd.cut(df['trestbps'] ,bins = [90 ,120,150,180,200] , 
                        labels = ['90-119' , '120-149' , '150-179' , '180-200'] , right = False)

trestbps_group_analysis = df.groupby('trestbps_group',observed = False).agg(total_patients = ('target' , 'count') ,disease_rate = ('target' , 'mean'))

trestbps_group_analysis['disease_rate'] *=100

print(trestbps_group_analysis . round(2))
```

```python
df['old_peak_group'] = pd.cut(df['oldpeak'] ,bins = [0,1,2,3,4,5,float('inf')] , 
                        labels = ['0-0.9','1-1.9' , '2-2.9' , '3-3.9' , '4-4.9' , '5+'] , right = False)

old_peak_group_analysis = df.groupby('old_peak_group',observed = False).agg(total_patients = ('target' , 'count') ,disease_rate = ('target' , 'mean'))

old_peak_group_analysis['disease_rate'] *=100

print(old_peak_group_analysis . round(2))
```

```python
sex_count = df['sex'].value_counts()

pie = plt.pie( sex_count , 
              labels = ['female' , 'male'] ,
              autopct = '%1.1f%%' ,
              colors = ['pink' , 'skyblue'] ,
              explode = [0.03 ,0] ,
              shadow = True ,
              startangle = 90)

plt.title('percentage of sex')
plt.axis('equal')
plt.show()
```

```python
sex_by_disease_rate = df.groupby('sex')['target'].mean()*100

plt.figure(figsize=(6,6))

bars=plt.bar(sex_by_disease_rate .index ,sex_by_disease_rate .values , color =['pink' , 'skyblue'])

plt.xticks([0,1] , ['female' , 'male'])
plt.xlabel('sex')
plt.ylabel('disease rate')
plt.title('sex by disease rate %')

for bar in bars : plt.text( bar.get_x() + bar.get_width() /2 ,
                           bar.get_height() ,
                           f"{bar.get_height():.1f}%",
                           ha = 'center' , va ='bottom')                   

plt.tight_layout()
plt.show()
```

```python
age_group_by_disease = df.groupby('Age_Group')['target'].mean()*100

bars =plt.bar(age_group_by_disease .index , age_group_by_disease. values , color = 'purple' )

plt.xlabel('age group')
plt.ylabel('disaes rate %')
plt.title('age group by disease rate')

for bar in bars : plt.text(bar.get_x() + bar.get_width() /2 ,
                           bar.get_height() ,
                           f'{bar.get_height() : .1f}%' ,
                           ha='center' , va ='bottom')

plt.tight_layout()
plt.show()
```
